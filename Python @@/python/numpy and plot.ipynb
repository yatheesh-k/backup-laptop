# plots in python  

```python
import matplotlib.pyplot as plt
plt.style.use("seaborn-whitegrid")
```

```python
x=[1,6,8]
y=[2,7,10]
```

```python
plt.figure(figsize=(18,6),facecolor="lightgreen")
plt.plot(x,y,color="darkblue",linewidth=4,linestyle=":")
plt.scatter(x,y,color="red",s=200,marker=">")
plt.xlabel("X Axis",color="red",size=20)
plt.ylabel("Y Axis",color="red",size=20)
plt.title("My First Plot in Python",color="red",size=23)


plt.xlim(0,11)
plt.ylim(0,11)
plt.xticks(range(0,11))
plt.yticks(range(0,11))
plt.savefig("image.png",dpi=100)
plt.show()
```

# merge of two data frames

```python
df1=pd.DataFrame({
    
    "name":["a","b","c","d"],
    "age":[19,20,30,29]
})


df2=pd.DataFrame({
    
    "name":["a","b","c","e"],
    "salary":[1,10,20,29]
})
```

```python
df1
```

```python
df2
```

```python
pd.merge(df1,df2,on="name",how="inner")
```

```python
pd.merge(df1,df2,on="name",how="outer")
```

```python
pd.merge(df1,df2,on="name",how="left")
```

```python
pd.merge(df1,df2,on="name",how="right")[["age","name",]]
```
